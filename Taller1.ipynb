```python
# 1. Registro de estudiantes
estudiantes = [
    ("Ana", 20, 4.5),
    ("Carlos", 22, 3.8),
    ("Laura", 19, 4.2),
    ("Diego", 21, 3.9),
    ("Sofía", 20, 4.1),
    ("Miguel", 23, 4.0),
    ("Valentina", 19, 3.7),
    ("Daniel", 22, 4.3)
]

print(estudiantes)
```

```python
# 2. Mostrar la información
for nombre, edad, nota in estudiantes:
    print(f"{nombre} tiene {edad} años y obtuvo una nota de {nota}")
```

```python
# 3. Clasificación de estudiantes
for nombre, edad, nota in estudiantes:
    if nota >= 4.5:
        clasificacion = "Excelente"
    elif nota >= 4.0:
        clasificacion = "Bueno"
    elif nota >= 3.0:
        clasificacion = "Aceptable"
    else:
        clasificacion = "Reprobó"

    print(f"{nombre}: {clasificacion}")


```

```python
# 4. Promedio general
suma_notas = 0

for nombre, edad, nota in estudiantes:
    suma_notas += nota  # Va acumulando cada nota

promedio = suma_notas / len(estudiantes)

print(f"El promedio general es: {promedio:.2f}")
```

```python
# 5. Busqueda de estudiante
nombre_buscado = "Laura"
encontrado = False

for estudiante in estudiantes:
    if estudiante[0] == nombre_buscado:
        print("Encontrado:", estudiante)
        encontrado = True
        break

if not encontrado:
    print("No se encontro el estudiante")
```

```python
# 6. Diccionario de ciudades
ciudades = {
    "Ana": "Bogotá",
    "Carlos": "Medellín",
    "Laura": "Bogotá",
    "Diego": "Cali",
    "Sofía": "Medellín",
    "Miguel": "Pereira",
    "Valentina": "Bogotá",
    "Daniel": "Cali"
}

for estudiante, ciudad in ciudades.items():
  print(f"{estudiante} vive en {ciudad}")
```

```python
# 7. Cantidad de estudiantes por ciudad
conteo_ciudades = {}

for ciudad in ciudades.values():
    if ciudad in conteo_ciudades:
        conteo_ciudades[ciudad] += 1
    else:
        conteo_ciudades[ciudad] = 1

print(conteo_ciudades)
```

```python
# 8. Ciclo While
cantidad = 0
suma = 0

while True:
    numero = float(input("Escribe un numero (0 para salir): "))
    if numero == 0:
        break
    cantidad = cantidad + 1
    suma = suma + numero

if cantidad > 0:
    promedio = suma / cantidad
    print("Cantidad de numeros:", cantidad)
    print("Suma total:", suma)
    print("Promedio de los numeros:", promedio)
else:
    print("No escribiste ningun numero")
```

```python
# 9. Uso de break y continue
for num in range(1, 31):
    if num == 25:
        break
    if num % 3 == 0:
        continue
    print(num)
```
